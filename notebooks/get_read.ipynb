### 取り回しにくいので生成AIにコードを作成してもらった

・各ディレクトリで実行しなければいけない
・標準出力なので、見落としてしまう可能性がある
・ファイルとの形式が結びつきにくい

```python
from pathlib import Path
import cv2
import numpy as np
import pandas as pd


def check_image_type(file_path):
    image = cv2.imread(str(file_path), cv2.IMREAD_UNCHANGED)

    if image is None:
        return "read_error"

    # 元から1チャンネル
    if image.ndim == 2:
        return "grayscale"

    # 3チャンネル以上
    if image.ndim == 3:
        b = image[:, :, 0]
        g = image[:, :, 1]
        r = image[:, :, 2]

        if np.array_equal(b, g) and np.array_equal(g, r):
            return "grayscale"
        else:
            return "color"

    return "unknown"


data_dir = Path("./data")

results = []

for file_path in data_dir.rglob("*.jpeg"):

    image_type = check_image_type(file_path)

    results.append({
        "split": file_path.parts[-3],   # train / test / val
        "label": file_path.parts[-2],   # NORMAL / PNEUMONIA
        "filename": file_path.name,
        "type": image_type
    })


df = pd.DataFrame(results)

print(df)
```

```python
from pathlib import Path
import cv2
import pandas as pd


def check_image_type(image):
    if image.ndim == 2:
        return "grayscale"

    return "color"


data_dir = Path("./data")

results = []

for file_path in data_dir.rglob("*.jpeg"):

    image = cv2.imread(
        str(file_path),
        cv2.IMREAD_UNCHANGED
    )

    if image is None:
        continue

    image_type = check_image_type(image)

    image_hist_cv = cv2.calcHist(
        [image],
        [0],
        None,
        [256],
        [0, 256]
    )

    results.append({
        "split": file_path.parts[-3],
        "label": file_path.parts[-2],
        "filename": file_path.name,
        "type": image_type,
        "hist": image_hist_cv.flatten()
    })

df = pd.DataFrame(results)

print(df)
```

```python
import numpy as np
import matplotlib.pyplot as plt

for (split, label), group in df.groupby(["split", "label"]):

    histograms = np.stack(group["hist"].to_numpy())

    mean_hist = histograms.mean(axis=0)

    plt.figure()
    plt.plot(range(256), mean_hist)

    plt.title(f"{split} / {label}")
    plt.xlabel("Pixel value")
    plt.ylabel("Average frequency")
    plt.ylim(0,20000)

    plt.show()
```

### 上記の結果から、グレースケールの画像であり背景色（黒が大半の画像になっていることが予測できる）

```python
from pathlib import Path
import cv2
import pandas as pd
import numpy as np

data_dir = Path("./data")

results = []

for file_path in data_dir.rglob("*.jpeg"):

    image = cv2.imread(
        str(file_path),
        cv2.IMREAD_UNCHANGED
    )

    # 読み込み失敗
    if image is None:
        results.append({
            "split": file_path.parts[-3],
            "label": file_path.parts[-2],
            "filename": file_path.name,
            "status": "broken"
        })
        continue

    # shape取得
    if image.ndim == 2:
        height, width = image.shape
        channels = 1
    else:
        height, width, channels = image.shape

    # 画素値統計
    min_pixel = image.min()
    max_pixel = image.max()
    mean_pixel = image.mean()
    std_pixel = image.std()

    # 異常判定
    status = "normal"

    if std_pixel == 0:
        status = "constant_image"

    elif height < 100 or width < 100:
        status = "too_small"

    results.append({
        "split": file_path.parts[-3],
        "label": file_path.parts[-2],
        "filename": file_path.name,
        "height": height,
        "width": width,
        "channels": channels,
        "min": min_pixel,
        "max": max_pixel,
        "mean": mean_pixel,
        "std": std_pixel,
        "status": status
    })


df = pd.DataFrame(results)

print(df)
```
